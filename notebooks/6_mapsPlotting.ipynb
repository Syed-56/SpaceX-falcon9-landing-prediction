*Completed:* Create and add a `folium.Circle` and `folium.Marker` for each launch site.

The four launch sites are marked on `site_map` using their latitude/longitude coordinates, with the site name shown in the popup and as a map label.

```python
# Initial the map
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# Add a circle and marker for each launch site.
site_colors = ['lightblue', 'lightgreen', 'black', 'orange']
site_labels = {
    'CCAFS LC-40': 'CCAFS Launch Complex 40',
    'CCAFS SLC-40': 'CCAFS Space Launch Complex 40',
    'KSC LC-39A': 'KSC Launch Complex 39A',
    'VAFB SLC-4E': 'VAFB Space Launch Complex 4E'
}

for i, row in launch_sites_df.iterrows():
    coordinate = [row['Lat'], row['Long']]
    label = site_labels.get(row['Launch Site'], row['Launch Site'])
    circle = folium.Circle(
        coordinate,
        radius=1500,
        color=site_colors[i % len(site_colors)],
        fill=True
    ).add_child(folium.Popup(label))
    marker = folium.map.Marker(
        coordinate,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>' % label
        )
    )
    site_map.add_child(circle)
    site_map.add_child(marker)

# Assign marker colors for Task 2
def assign_marker_color(launch_class):
    return 'green' if launch_class == 1 else 'red'

spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

site_map
```

Next, let's create markers for all launch records.
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

*Completed:* Create a new `marker_color` column in `spacex_df` based on the `class` value.

- `class = 1` → `green` (successful launch)
- `class = 0` → `red` (failed launch)


```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
# Task 2: Mark the success/failed launches for each site on the map
site_map.add_child(marker_cluster)

for index, record in spacex_df.iterrows():
    coordinate = [record['Lat'], record['Long']]
    marker = folium.Marker(
        coordinate,
        icon=folium.Icon(color='white', icon_color=record['marker_color'])
    )
    marker_cluster.add_child(marker)

site_map
```

```python
# Add marker_cluster to current site_map
site_map.add_child(marker_cluster)

# Add one marker for each launch record.
# Green indicates success (class=1); red indicates failure (class=0).
for index, record in spacex_df.iterrows():
    coordinate = [record['Lat'], record['Long']]
    marker = folium.Marker(
        coordinate,
        icon=folium.Icon(color='white', icon_color=record['marker_color'])
    )
    marker_cluster.add_child(marker)

site_map
```

Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*Completed:* For the VAFB SLC-4E example, use the closest coastline point identified with `MousePosition` and calculate the distance.

Closest coastline point used by this notebook example:
- Latitude: **34.63462**
- Longitude: **-120.62439**
- Distance from VAFB SLC-4E: **1.26 km**


```python
# Coordinates of the selected closest coastline point for VAFB SLC-4E
coastline_lat = 34.63462
coastline_lon = -120.62439

launch_site_lat = launch_sites_df.loc[launch_sites_df['Launch Site'] == 'VAFB SLC-4E', 'Lat'].iloc[0]
launch_site_lon = launch_sites_df.loc[launch_sites_df['Launch Site'] == 'VAFB SLC-4E', 'Long'].iloc[0]

distance_coastline = calculate_distance(
    launch_site_lat, launch_site_lon,
    coastline_lat, coastline_lon
)

print(f"Distance to coastline: {distance_coastline:.2f} km")
```

```python
# Add a marker at the selected coastline point and display the distance.
coastline_coordinate = [coastline_lat, coastline_lon]

distance_marker = folium.Marker(
    coastline_coordinate,
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>'
              % "{:10.2f} KM".format(distance_coastline),
    )
)
```

*Completed:* Draw a `PolyLine` between VAFB SLC-4E and the selected coastline point.


```python
# Draw a line from the VAFB SLC-4E launch site to the selected coastline point.
launch_site_coordinate = [launch_site_lat, launch_site_lon]
lines = folium.PolyLine(
    locations=[coastline_coordinate, launch_site_coordinate],
    weight=2
)
site_map.add_child(lines)
site_map.add_child(distance_marker)
```

```python
# Create markers with distances to the closest railway, highway, and city.
# Draw a line from each proximity point to VAFB SLC-4E.
railway_coordinates = coastline_coordinate
highway_coordinates = (34.68577, -120.59585)
city_coordinates = (34.63886, -120.45788)

distance_railway = distance_coastline
distance_highway = calculate_distance(
    launch_site_lat, launch_site_lon,
    highway_coordinates[0], highway_coordinates[1]
)
distance_city = calculate_distance(
    launch_site_lat, launch_site_lon,
    city_coordinates[0], city_coordinates[1]
)

dist_railway_marker = folium.Marker(
    railway_coordinates,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>'
              % "{:10.2f} KM".format(distance_railway),
    )
)

dist_highway_marker = folium.Marker(
    highway_coordinates,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>'
              % "{:10.2f} KM".format(distance_highway),
    )
)

dist_city_marker = folium.Marker(
    city_coordinates,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>'
              % "{:10.2f} KM".format(distance_city),
    )
)
```
